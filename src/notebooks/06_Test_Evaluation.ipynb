```python
# Configuration
MODEL_PATH = '../runs/detect/compound_FINAL_medium_1280_FINETUNE/weights/best.pt'  # adjust to your weights
DATA_YAML = '../../dataset/04_all_classes/data.yaml'  # or 05/06 dataset yaml
CONF_THRES = 0.001  # low conf for metrics
IOU_THRES = 0.6     # default
IMG_SIZE = 1280      # match training resolution
DEVICE = 0          # set to 'cpu' for CPU, or index for GPU
```

```python
from ultralytics import YOLO

model = YOLO(MODEL_PATH)
model.overrides['conf'] = CONF_THRES
model.overrides['iou'] = IOU_THRES
model.overrides['imgsz'] = IMG_SIZE

# Run validation on test split
results = model.val(data=DATA_YAML, split='test', device=DEVICE)
results
```

## Tips
- Set `MODEL_PATH` to your checkpoint (e.g., runs/detect/.../best.pt).
- Choose the appropriate `data.yaml` (04_all_classes, 05_selected_classes, or 06_compound_chart_splitter).
- For CPU-only, set `DEVICE = 'cpu'` (will be slower).
- Metrics printed here are for the **test** split.
